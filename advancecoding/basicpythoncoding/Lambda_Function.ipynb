# Lambda Function

1. **Anonymous** → Means a **nameless function**
2. **Inline** → Written directly where it is needed
3. **Single-use function** → Usually used for a small, simple operation
4. **Returns one value** → No `return` keyword is required
5. **Used as input to Higher-Order Functions**

### Higher-Order Function

A **Higher-Order Function** is a function that **accepts another function as an argument**.

Examples:

- `map()`
- `filter()`
- `reduce()`

### Example

```python
nums = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, nums))

print(result)

# 🗺️ Python `map()`, `filter()`, and `reduce()`

## Mind Map

```text
Functional Programming
│
├── map()    → Transform every item
├── filter() → Keep selected items
└── reduce() → Combine items into one result

## 1. `map()` → Transform

### Syntax

`map(function, iterable)`

### Example

```python
nums = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, nums))

print(result)

Output

[2, 4, 6, 8]

Best case: When you want to change every item.

## 2. `filter()` → Select

### Syntax

`filter(function, iterable)`

### Example

```python
nums = [1, 2, 3, 4]

result = list(filter(lambda x: x % 2 == 0, nums))

print(result)

Output

[2, 4]

Best case: When you want to keep only matching items.

## 3. `reduce()` → Combine

### Syntax

`reduce(function, iterable)`

## `reduce()` → Two Parameters

`reduce()` normally uses a lambda function with **two parameters**.

### Syntax

`reduce(lambda a, b: operation, iterable)`


### Example

```python
from functools import reduce

nums = [1, 2, 3, 4]

result = reduce(lambda a, b: a + b, nums)

print(result)

Output

10

Best case: When you want to combine all items into one value
````

### Example reduce() → Two Parameters

```python
from functools import reduce

result = reduce(lambda a, b: a + b, [1, 2, 3, 4])

print(result)

`````python
# some practice examples 

print("-----------Map----------------------")
res=list(map(lambda a  :  a%2==0, [i for i in range(10)] ))
print(res)
print("-----------Filter----------------------")
res=list(filter(lambda a  :  a%2==0, [i for i in range(10)] ))
print(res)


`````

`````python
print(
    "----------- Reduce -----------\n"
    "Key note: reduce returns one value only.\n"
    "So, don't assign it to a list.\n"
    "It does not return a list of values like map and filter."
)
print("")

from functools import reduce
nlist=[1,2,3]
res=reduce(lambda a,b: a+b, nlist)
print(res)
`````

## Simple rule
```
map()    → lambda usually takes 1 parameter → lambda x: ...
filter() → lambda usually takes 1 parameter → lambda x: ...
reduce() → lambda usually takes 2 parameters → lambda a, b: ...
