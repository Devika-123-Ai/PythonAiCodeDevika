# Functions

```python
## add function

def add_function():
    print("Enter numbers for addition")
    num1=int(input("Enter the first number"))
    num2=int(input("Enter the second number"))
    print("Result: ",num1+num2)

add_function()
```

```python
## generate the even numbers from the given number range

def get_even_numbers():
    print("Print Even Numbers")
    num=int(input("Enter the number range: "))
    evenList=[n for n in range(num) if n%2==0]
    print(evenList)
  #get_even_numbers()  # way of calling function

## assign the function address to variable
# now address of this function is assigned to x so you can call the x to run the get_even_numbers
x = get_even_numbers
x()

    
```

## Function with Params

```python
## Function with return value

def get_cube_function(num):
    return num*num*num

print(get_cube_function(2))

         
```

```python
## Example: String + Integer
def greet(name, age):
    print("Hello", name)
    print("You are", age, "years old")

greet("John", 25)
```

```python
## Another example: String + two numbers
def add_numbers(name, num1, num2):
    total = num1 + num2
    print(name, "the total is", total)

add_numbers("Alice", 10, 20)
```

```python
## You can also use type hints to make the expected types clear:

def add_numbers(name: str, num1: int, num2: int):
    total = num1 + num2
    print(name, "the total is", total)

add_numbers("Alice", 10, 20)
```
