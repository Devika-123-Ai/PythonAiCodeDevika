```python
def numbers():
    yield 10
    yield 20
    yield 30

g = numbers()       #A generator is a special type of iterator that produces values one at a time using yield.

print(next(g))
print(next(g))
print(next(g))
```

```python
names = ["John", "Sam", "Mike"]             #for loop without next()


for index, name in enumerate(names):    #**`enumerate()`** → gives you both index and value while looping.
    print(index, name)
```

```python
names = ["John", "Sam", "Mike"]

e = enumerate(names)        #enumerate(names) returns an iterator, so you can use next() on it.

print(next(e))
print(next(e))
print(next(e))
```

```python
# What is return inside a generator?

# In a normal function:

def add():
    return 10

#return gives the result and ends the function.



def numbers():
    yield 10
    yield 20        #return means:in generators Stop the generator completely.
    return
    yield 30
    g = numbers()

print(next(g))  # 10
print(next(g))  # 20
print(next(g))  # StopIteration
```

```python
#iterable-container -holds the data like tuple ,list ,dict
#iterator-tracks the position using iter()
my_list=[10,20,30]  #iterable
my_iter = iter(my_list)     #iterator-track the position 
print(next(my_iter))        #iter()-creates iterator
                            #next()-get the next value from iterator
print(next(my_iter))    #An iterator is an object that lets us access elements one by one using the next() function.”
```
