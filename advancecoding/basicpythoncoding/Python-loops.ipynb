#### While loop

```python
## print numbers 1 to 10
x=1
while x<=10:
    print(x)
    x=x+1
    
## print numbers in a series in decreasing order 
num=int(input("Enrer the number to print series."))
y=num
while y>=1:
    print(y)
    y=y-1
```

## For loop

```python
import time
```

```python
# it is designed to extract the values from sequentilal datatypes
# example

fruits=["APPLE", "BANANA","MANGO","GRAPES","CHERRY"]
for x in fruits:
    time.sleep(1)
    if x!="BANANA":
        print(x)
        
        
```

```python
s="Beautiful"
for x in s:
    print(x, end = " ")
    time.sleep(1)
    
```
