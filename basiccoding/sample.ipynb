```python
#Normal function 
def add(a,b):
    return(a+b)
print(10+20)
```

```python
#Lambda function: A small, anonymous function defined using lambda, usually for a simple one-line operation.

add = lambda a, b: a + b 
print(add(10, 20))   #—————>single line statements
```

```python
#Lambda with filter()
##filter() is used to select items that satisfy a condition.
#Example: Get only even numbers:
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
```


Lambda functions are especially useful when you need a small temporary function, for example with sorted(), map(), or filter().

Example with sorting:


```python
students = [
    ("Ravi", 80),
    ("John", 95),
    ("Sam", 75)
]

students.sort(key=lambda x: x[1])

print(students)
```

```python
# Lambda with map()
# map() is used to apply a function to every item in a collection.
# Example: Multiply every number by 2:


numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x * 2, numbers)

print(list(result))
```
