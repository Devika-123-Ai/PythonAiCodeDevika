# 📝 Python List Operations (CRUD)

A **List** is a mutable, ordered collection that can store multiple values (including different data types) in a single variable.

---

# 🟢 1. CREATE

## Definition

Creating a list means storing multiple values inside a single variable.

---

## 📋 Create Summary Table

| Method | Purpose | Syntax | Example | Output |
|---------|---------|--------|---------|--------|
| Empty List | Create an empty list | `list_name = []` | `numbers = []` | `[]` |
| Empty List Constructor | Create an empty list | `list_name = list()` | `numbers = list()` | `[]` |
| Manual List | Create with predefined values | `list_name = [item1, item2]` | `fruits = ["Apple","Banana"]` | `['Apple','Banana']` |
| Convert Iterable | Convert string, tuple, set, etc. | `list(iterable)` | `list("Python")` | `['P','y','t','h','o','n']` |
| List Comprehension | Create using loop | `[expression for item in iterable]` | `[x*x for x in range(5)]` | `[0,1,4,9,16]` |

---

## Examples

### Empty List

```python
numbers = []
print(numbers)
```

Output

```text
[]
```

---

### Empty List using Constructor

```python
numbers = list()
print(numbers)
```

Output

```text
[]
```

---

### Manual List

```python
fruits = ["Apple", "Banana", "Orange"]
print(fruits)
```

Output

```text
['Apple', 'Banana', 'Orange']
```

---

### Convert String to List

```python
letters = list("Python")
print(letters)
```

Output

```text
['P', 'y', 't', 'h', 'o', 'n']
```

---

### List Comprehension

```python
numbers = [x**2 for x in range(6)]
print(numbers)
```

Output

```text
[0, 1, 4, 9, 16, 25]
```

---

## 🧠 Create Mind Map

```text
Create
│
├── []
├── list()
├── Manual List
├── list(iterable)
└── List Comprehension
```

---

# 🔵 2. READ

## Definition

Reading means accessing elements stored inside a list.

---

## 📋 Read Summary Table

| Method | Purpose | Syntax | Example |
|---------|----------|--------|---------|
| Indexing | Access one element | `list[index]` | `numbers[0]` |
| Slicing | Access multiple elements | `list[start:stop:step]` | `numbers[1:5:2]` |
| for Loop | Read all elements | `for item in list:` | `for x in numbers:` |
| enumerate() | Read index and value | `enumerate(list)` | `for i,v in enumerate(numbers)` |

---

# 1️⃣ Indexing

### Syntax

```python
list[index]
```

### Example

```python
fruits = ["Apple", "Banana", "Orange"]

print(fruits[0])
print(fruits[-1])
```

Output

```text
Apple
Orange
```

---

# 2️⃣ Slicing

### Syntax

```python
list[start : stop : step]
```

### Parameters

| Parameter | Meaning |
|-----------|---------|
| start | Starting index (inclusive) |
| stop | Ending index (exclusive) |
| step | Number of positions to move |

---

### Slicing Examples

```python
numbers = [10,20,30,40,50,60,70,80]
```

| Expression | Result |
|------------|---------|
| `numbers[:]` | `[10,20,30,40,50,60,70,80]` |
| `numbers[2:]` | `[30,40,50,60,70,80]` |
| `numbers[:4]` | `[10,20,30,40]` |
| `numbers[1:5]` | `[20,30,40,50]` |
| `numbers[::2]` | `[10,30,50,70]` |
| `numbers[1:7:2]` | `[20,40,60]` |
| `numbers[::-1]` | `[80,70,60,50,40,30,20,10]` |

---

# 3️⃣ Using for Loop

### Syntax

```python
for item in list:
    print(item)
```

### Example

```python
fruits = ["Apple","Banana","Orange"]

for fruit in fruits:
    print(fruit)
```

---

# 4️⃣ Using enumerate()

### Syntax

```python
for index, value in enumerate(list):
```

### Example

```python
fruits = ["Apple","Banana","Orange"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

Output

```text
0 Apple
1 Banana
2 Orange
```

---

## 📊 Indexing vs Slicing vs Loop

| Feature | Indexing | Slicing | Loop |
|----------|----------|----------|------|
| Returns | One element | Multiple elements | One by one |
| Uses Index | ✅ | ✅ | Optional |
| Returns List | ❌ | ✅ | ❌ |
| Iterate All | ❌ | ❌ | ✅ |

# 📦 Python Tuple

---

# 📚 What is a Tuple?

A **Tuple** is an **ordered, immutable collection** that can store multiple heterogeneous values.

### Example

```python
student = (101, "John", 95.5, True)
```

---

# ⭐ Tuple Properties

- ✅ Ordered
- ✅ Immutable (Cannot be modified)
- ✅ Duplicates Allowed
- ✅ Heterogeneous Data
- ✅ Indexing Supported
- ✅ Slicing Supported
- ✅ Nesting Supported
- ✅ Faster than List

---

# 📋 Tuple Summary

| Feature | Description |
|----------|-------------|
| Syntax | `( )` |
| Ordered | ✅ Yes |
| Mutable | ❌ No |
| Duplicate Values | ✅ Yes |
| Indexing | ✅ Yes |
| Slicing | ✅ Yes |
| Nested | ✅ Yes |

---

# 🟢 Create Tuple

| Method | Syntax | Example |
|---------|--------|---------|
| Empty Tuple | `()` | `t = ()` |
| Constructor | `tuple()` | `t = tuple()` |
| Manual | `(1,2,3)` | `t=(10,20,30)` |
| Convert Iterable | `tuple(iterable)` | `tuple("Python")` |

### Example

```python
numbers = (10,20,30)
```

---

# 🔵 Read Tuple

| Method | Syntax | Example |
|----------|--------|---------|
| Indexing | `t[index]` | `t[0]` |
| Slicing | `t[start:stop:step]` | `t[1:4]` |
| Loop | `for x in t:` | Read all values |

### Example

```python
t = (10,20,30,40)

print(t[1])
print(t[1:3])
```

---

# 🟡 Update Tuple

❌ Tuples **cannot be updated** because they are immutable.

```python
t = (10,20)

t[0] = 100
```

Output

```text
TypeError
```

---

# 🔴 Delete Tuple

| Operation | Syntax |
|-----------|--------|
| Delete Entire Tuple | `del tuple_name` |

Example

```python
t = (10,20)

del t
```

# 🚀 Python List Comprehension

---

# 📚 What is List Comprehension?

**List comprehension** is a compact way to create a new list by applying an expression on each element of an iterable.

It is a shorter and cleaner alternative to writing a `for` loop.

---

# 🧠 Basic Idea

### Traditional Way

```python
numbers = []

for i in range(5):
    numbers.append(i)

print(numbers)
```

Output

```text
[0,1,2,3,4]
```

---

### Using List Comprehension

```python
numbers = [i for i in range(5)]

print(numbers)
```

Output

```text
[0,1,2,3,4]
```

---

# ⭐ Why Use List Comprehension?

| Benefit | Explanation |
|---------|-------------|
| Less Code | Reduces multiple lines into one expression |
| Readable | Easy to understand simple transformations |
| Faster | Usually faster than normal loops |
| Pythonic | Preferred Python coding style |
| Creates New List | Generates a new list easily |

---

# 📌 When to Use List Comprehension?

Use list comprehension when:

✅ You need to create a new list  
✅ You are transforming existing data  
✅ You are filtering data  
✅ Logic is simple and readable  
✅ You are applying the same operation to every element  

---

# ❌ When NOT to Use List Comprehension?

Avoid list comprehension when:

❌ Logic is complex  
❌ Multiple conditions make it hard to read  
❌ You need multiple statements  
❌ You are modifying existing data  

Example:

Avoid:

```python
[
    process(x)
    for x in data
    if condition1
    if condition2
    if condition3
]
```

Use a normal loop instead.

---

# 📋 List Comprehension Syntax

## Basic Syntax

```python
[expression for item in iterable]
```

---

## Syntax Breakdown

| Part | Meaning |
|------|---------|
| expression | Value to store in list |
| item | Each element from iterable |
| iterable | Collection to loop through |

---

# 🟢 1. Basic List Comprehension

## Syntax

```python
[expression for item in iterable]
```

---

## Example

```python
numbers = [x for x in range(5)]

print(numbers)
```

Output

```text
[0,1,2,3,4]
```

---

# 🟡 2. List Comprehension with Expression

Modify each element while creating list.

## Example

Create squares:

```python
numbers = [x*x for x in range(6)]

print(numbers)
```

Output

```text
[0,1,4,9,16,25]
```

---

# 🔵 3. List Comprehension with Condition

Filter values using `if`.

## Syntax

```python
[expression for item in iterable if condition]
```

---

## Example

Get even numbers:

```python
even = [x for x in range(10) if x % 2 == 0]

print(even)
```

Output

```text
[0,2,4,6,8]
```

---

# 🟣 4. List Comprehension with if-else

## Syntax

```python
[value_if_true if condition else value_if_false for item in iterable]
```

---

## Example

Replace numbers:

```python
result = [
    "Even" if x % 2 == 0 else "Odd"
    for x in range(5)
]

print(result)
```

Output

```text
['Even','Odd','Even','Odd','Even']
```

---

# 🟠 5. Nested List Comprehension

Creating lists inside lists.

## Syntax

```python
[
 expression
 for item1 in iterable1
 for item2 in iterable2
]
```

---

## Example

Create multiplication pairs:

```python
pairs = [
    (x,y)
    for x in [1,2]
    for y in [3,4]
]

print(pairs)
```

Output

```text
[(1,3),(1,4),(2,3),(2,4)]
```

# 🚀 List Comprehension Execution Order

## Syntax

```python
[ output_expression for item in input_sequence if condition ]
```

---

## Execution Flow (Order)

```text
[ output_expression  for item in input_sequence  if condition ]

          3️⃣                 1️⃣                 2️⃣
          │                  │                  │
          ▼                  ▼                  ▼

       Create value     Take each item     Check condition
```

---

## Actual Execution Order

```text
Step 1️⃣ → Read values from input sequence

Step 2️⃣ → Apply condition (if present)

Step 3️⃣ → Execute output expression

Step 4️⃣ → Store result in new list
```

---

## Example

```python
squares = [x*x for x in range(10) if x % 2 == 0]
```

### Execution:

```
range(10)
    │
    ▼
Step 1️⃣ Get each x

x = 0
x = 1
x = 2
x = 3
...

    │
    ▼
Step 2️⃣ Check condition

x % 2 == 0

    │
    ▼
Step 3️⃣ Calculate expression

x*x

    │
    ▼
Step 4️⃣ Store result

[0,4,16,36,64]
```

---

# 🧠 Easy Memory Trick

```text
FOR → IF → OUTPUT

1️⃣ for
   Get the data

2️⃣ if
   Filter the data

3️⃣ expression
   Create the result
```

---

# Visual Syntax Map

```text
[ expression  for item in sequence  if condition ]

      3️⃣              1️⃣               2️⃣
      │               │                │
      ▼               ▼                ▼

   WHAT TO ADD    WHERE TO GET     WHAT TO KEEP
```
