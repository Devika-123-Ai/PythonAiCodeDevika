## Read List 
 ### 1. indexing
 ### 2. slicing 
 ### 3. loop
 ### 4. enumurate

```python
fruits=["Apple", "Banana", "orange"] ##indexing
print(fruits[0])
print(fruits[-1])

print("--------------- Slicing -----------------------------------")
print(fruits[0:1:1])
print("--------------- for loop -----------------------------------")
for fruit in fruits:
    print(fruit)
print("--------------- enumurate-----------------------------------")
for index, fruit in enumerate(fruits):
    print(index,fruit)

```

## Update List

```python
fruits=["Apple", "Banana", "orange"]
```

```python
fruits.append("kiwi") # append
print(fruits)
```

```python
fruits.extend(["cherry","mango"]) ## extend
print(fruits)
```

```python
fruits.insert(1,"pineapple") # insert
print(fruits)
```

```python
fruits[1]="dragon" ## replace based on index
print(fruits)
```

## Delete

```python
fruits=["Apple","Banana","orange","cherry"]
print(fruits)
```

```python
fruits.remove("cherry")
print(fruits)
```

```python
print(fruits.pop()) ##pop without index
print(fruits)
```

```python
fruits.pop(0) ##pop with index
print(fruits)
```

```python
fruits.extend(["Apple","cherry","Banana"])
print(fruits)
```

```python
del fruits[1]
print(fruits)
```

```python
del fruits ## del entire fruits variable
print(fruits)
```
